```python
# ==========================================
# AUTOMATIC SPEECH RECOGNITION SYSTEM
# Using Whisper + Gradio
# ==========================================

!pip -q install openai-whisper gradio

import whisper
import gradio as gr

# Load Whisper model
model = whisper.load_model("base")

# Speech recognition function
def speech_to_text(audio):
    if audio is None:
        return "Please upload or record an audio file."

    try:
        result = model.transcribe(audio)
        return result["text"].strip()

    except Exception as e:
        return f"Error: {str(e)}"


# Create Gradio interface
with gr.Blocks(title="Automatic Speech Recognition") as app:

    gr.Markdown("""
    # 🎙️ Automatic Speech Recognition System
    ### Convert spoken audio into text using AI
    """)

    audio_input = gr.Audio(
        sources=["upload", "microphone"],
        type="filepath",
        label="🎤 Upload or Record Speech"
    )

    output_text = gr.Textbox(
        label="📝 Recognized Text",
        lines=6,
        placeholder="Your speech will appear here..."
    )

    button = gr.Button("🔊 Convert Speech to Text")

    button.click(
        fn=speech_to_text,
        inputs=audio_input,
        outputs=output_text
    )

    gr.Markdown("""
    **Technology:** Python + Whisper + Gradio
    **Application:** Automatic Speech Recognition (ASR)
    """)


# Launch application
app.launch(share=True)
```
